```python
import os

image_dir = '/home/mobilion/Desktop/dataset/yoloformat/pp1'
label_dir = '/home/mobilion/Desktop/dataset/yoloformat/pp1'

images = {os.path.splitext(f)[0] for f in os.listdir(image_dir) if f.endswith(('.jpg', '.png'))}
labels = {os.path.splitext(f)[0] for f in os.listdir(label_dir) if f.endswith('.txt')}

# 서로 일치하지 않는 이미지와 라벨 파일 찾기
unmatched_images = images - labels
unmatched_labels = labels - images

print(f"Unmatched images: {unmatched_images}")
print(f"Unmatched labels: {unmatched_labels}")

# 불일치하는 파일이 있는 경우 이들을 확인하여 삭제하거나 수정
```

```python
import os

# 이미지와 라벨 파일이 있는 폴더 경로를 지정하세요.
image_dir = '/home/mobilion/Desktop/dataset/yoloformat/pp1'  # 이미지 파일이 있는 폴더 경로
label_dir = '/home/mobilion/Desktop/dataset/yoloformat/pp1'  # 라벨 파일이 저장될 폴더 경로

# 이미지 폴더 내의 모든 .png 파일 검색
for filename in os.listdir(image_dir):
    if filename.endswith('.png'):
        # 이미지 파일 이름에서 확장자를 제거하여 기본 파일명 추출
        base_name = os.path.splitext(filename)[0]

        # 해당 이미지에 대응하는 .txt 파일 경로 설정
        txt_file_path = os.path.join(label_dir, f"{base_name}.txt")

        # .txt 파일이 없을 경우 빈 파일 생성
        if not os.path.exists(txt_file_path):
            with open(txt_file_path, 'w') as f:
                # 빈 파일 생성 (필요 시 기본 내용 추가 가능)
                pass
            print(f"Created empty label file: {txt_file_path}")

print("All missing .txt files have been created.")
```
